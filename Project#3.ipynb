```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
import datetime

# 1. Connect to PostgreSQL (Update with your credentials)
# Format: postgresql://username:password@host:port/database_name
engine = create_engine('postgresql://your_username:your_password@localhost:5432/your_database')

# 2. Simulate Master Data
warehouses = pd.DataFrame({
    'warehouse_id': ['WH01'],
    'warehouse_name': ['Central Distribution'],
    'location': ['Chicago, IL']
})

bins = pd.DataFrame({
    'bin_id': [f'A-{i}' for i in range(1, 11)], # Bins A-1 through A-10
    'warehouse_id': ['WH01'] * 10,
    'bin_type': ['High Rack'] * 10,
    'max_capacity_qty': [500] * 10
})

# 3. Simulate Transactional Data (Receiving & Putaway)
np.random.seed(42) # For reproducible results
dates = pd.date_range(start="2023-10-01", end="2023-10-31")

receipts_data = []
for _ in range(50): # 50 inbound deliveries
    receipts_data.append({
        'material_id': np.random.choice(['MAT-100', 'MAT-200', 'MAT-300']),
        'receipt_date': np.random.choice(dates),
        'qty_received': np.random.randint(50, 200),
        'destination_bin_id': np.random.choice(bins['bin_id'])
    })
df_receipts = pd.DataFrame(receipts_data)

# 4. Simulate Transactional Data (Picking & Shipping)
issues_data = []
for _ in range(30): # 30 outbound shipments
    issues_data.append({
        'material_id': np.random.choice(['MAT-100', 'MAT-200', 'MAT-300']),
        'issue_date': np.random.choice(dates),
        'qty_issued': np.random.randint(10, 50),
        'source_bin_id': np.random.choice(bins['bin_id'])
    })
df_issues = pd.DataFrame(issues_data)

# 5. Load Data to PostgreSQL (The 'L' in ETL)
warehouses.to_sql('warehouse', engine, if_exists='append', index=False)
bins.to_sql('storage_bins', engine, if_exists='append', index=False)
df_receipts.to_sql('goods_receipts', engine, if_exists='append', index=False)
df_issues.to_sql('goods_issues', engine, if_exists='append', index=False)

print("Warehouse simulation data successfully loaded to PostgreSQL!")
```
