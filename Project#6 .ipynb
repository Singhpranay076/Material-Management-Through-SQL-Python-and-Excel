```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

# 1. Connect to Database (Update with your credentials)
engine = create_engine('postgresql://your_username:your_password@localhost:5432/your_database')

# ==========================================
# EXTRACTION & TRANSFORMATION (The 'E' & 'T')
# ==========================================

# 2. Generate Date Dimension (A classic Data Engineering task)
start_date = '2023-01-01'
end_date = '2023-12-31'
dates = pd.date_range(start=start_date, end=end_date)

dim_date = pd.DataFrame({
    'date_key': dates.strftime('%Y%m%d').astype(int),
    'full_date': dates,
    'day_of_week': dates.day_name(),
    'month_name': dates.month_name(),
    'quarter': dates.quarter,
    'year': dates.year
})

# 3. Simulate Extracted Master Data
dim_material = pd.DataFrame({
    'material_key': [1, 2, 3],
    'material_id': ['MAT-10', 'MAT-20', 'MAT-30'],
    'material_description': ['Mountain Frame', 'Road Tire', 'Steel Chain'],
    'material_type': ['Assembly', 'Raw Material', 'Raw Material'],
    'standard_cost': [150.00, 25.50, 12.00]
})

dim_vendor = pd.DataFrame({
    'vendor_key': [1, 2],
    'vendor_id': ['V-100', 'V-200'],
    'vendor_name': ['Global Metals Inc.', 'RubberTech Co.'],
    'vendor_country': ['USA', 'Germany']
})

dim_plant = pd.DataFrame({
    'plant_key': [1],
    'plant_id': ['PL-1000'],
    'plant_location': ['Chicago, IL']
})

# 4. Simulate Transformed Fact Data
# In reality, you would merge your raw CSVs/Tables with the dimension tables to get the 'keys'
fact_procurement = pd.DataFrame({
    'date_key': [20231001, 20231015, 20231102],
    'material_key': [1, 2, 3],
    'vendor_key': [1, 2, 1],
    'plant_key': [1, 1, 1],
    'po_quantity': [50, 200, 500],
    'total_spend': [7500.00, 5100.00, 6000.00],
    'lead_time_days': [14, 7, 5]
})

# ==========================================
# LOAD (The 'L')
# ==========================================

# 5. Load to PostgreSQL Data Warehouse
# We use if_exists='append' because the tables are already created with relationships
dim_date.to_sql('dw_dim_date', engine, if_exists='append', index=False)
dim_material.to_sql('dw_dim_material', engine, if_exists='append', index=False)
dim_vendor.to_sql('dw_dim_vendor', engine, if_exists='append', index=False)
dim_plant.to_sql('dw_dim_plant', engine, if_exists='append', index=False)
fact_procurement.to_sql('dw_fact_procurement', engine, if_exists='append', index=False)

print("Data Warehouse ETL pipeline executed successfully!")
```
